```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
```

```python
df= pd.read_csv("titanic.csv")
df.head()
```

```python
df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin', 'SibSp', 'Embarked', 'Parch'], axis=1, inplace=True)
```

```python
df.head()
```

```python
target = df['Survived']
inputs= df.drop('Survived', axis='columns')
```

```python
inputs['Sex']= LabelEncoder().fit_transform(inputs['Sex'])
inputs.head()
```

```python
inputs.columns[inputs.isna().any()]
```

```python
inputs['Age'].fillna(inputs['Age'].mean(), inplace=True)
```

```python
inputs.columns[inputs.isna().any()]
```

```python
x_train, x_test, y_train, y_test = train_test_split(inputs, target, test_size=0.2)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
x_train.head()
y_train.head()
```

```python
titanic_model = GaussianNB()
titanic_model.fit(x_train, y_train)
```

```python
titanic_model.predict([[22, 1, 0, 7.25]])
```

```python
titanic_model.score(x_test, y_test)
```

```python
df1= pd.read_csv("Spam.csv")
df1.head()
```

```python
df1.groupby('Category').describe()
```

```python
df1['spam'] = df1['Category'].apply(lambda x: 1 if x == 'spam' else 0)
df1.head()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(df1['Message'], df1['spam'], test_size=0.25)
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X_train_cv = cv.fit_transform(X_train)
X_train_cv.toarray()[:3]
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_model = MultinomialNB()
spam_model.fit(X_train_cv, Y_train)
```

```python
emails= ['Hey mohan, can we get together to watch footbal game tomorrow?',
    'Upto 20% discount on parking, exclusive offer just for you. Dont miss this reward!']
emails_count = cv.transform(emails)
spam_model.predict(emails_count)
```

```python
X_test_count = cv.transform(X_test)
spam_model.score(X_test_count, Y_test)
```

```python
from sklearn.pipeline import Pipeline
spam_pipeline = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('NBv', MultinomialNB())
])
```

```python
spam_pipeline.fit(X_train, Y_train)
```

```python
spam_pipeline.predict(emails)
```

```python
spam_pipeline.score(X_test, Y_test)
```

```python
from sklearn.datasets import load_wine
wine= load_wine()
wine_df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
wine_df.head()
wine_df['target'] = wine.target
wine_df
```

```python
xx= wine_df.drop('target', axis='columns')
yy= wine_df['target']
```

```python
xx_train, xx_test, yy_train, yy_test = train_test_split(xx, yy, test_size=0.2)
xx_train.shape, xx_test.shape, yy_train.shape, yy_test.shape 
```

```python
wine_model_guassian = GaussianNB()
wine_model_guassian.fit(xx_train, yy_train)
wine_model_guassian.predict([[13.5, 2.5, 2.0, 20.0, 100.0, 2.5, 2.0, 0.3, 1.0, 3.0, 1.0, 3.0, 1000]])
```

```python
wine_model_guassian.score(xx_test, yy_test)
```

```python
wine_model_multinomial = MultinomialNB()
wine_model_multinomial.fit(xx_train, yy_train)
wine_model_multinomial.predict([[13.5, 2.5, 2.0, 20.0, 100.0, 2.5, 2.0, 0.3, 1.0, 3.0, 1.0, 3.0, 1000]])
```

```python
wine_model_multinomial.score(xx_test, yy_test)
```
