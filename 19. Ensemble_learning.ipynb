```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import SVC
```

```python
df= pd.read_csv("diabetes.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.Outcome.value_counts()
```

```python
x= df.drop("Outcome", axis=1)
y= df["Outcome"]
```

```python
scaler= StandardScaler()
x_scaled= scaler.fit_transform(x)
x_scaled
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2,stratify=y, random_state=10)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
scores= cross_val_score(DecisionTreeClassifier(), x_train, y_train, cv=5)
scores
```

```python
scores.mean()
```

```python
bag_model= BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=100,oob_score=True, max_samples=0.8, random_state=0)
bag_model.fit(x_train, y_train)
bag_model.oob_score_
```

```python
bag_model= BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=100,oob_score=True, max_samples=0.8, random_state=0)
scores1= cross_val_score(bag_model, x_train, y_train, cv=5).mean()
scores1
```

```python
scoresrdf= cross_val_score(RandomForestClassifier(), x_train, y_train, cv=5).mean()
scoresrdf
```

```python

df3 = df2[df2.Oldpeak<=(df2.Oldpeak.mean()+3*df2.Oldpeak.std())]
df3.shape
```

```python
df1[df1.RestingBP>(df1.RestingBP.mean()+3*df1.RestingBP.std())]
```

```python
df4 = df3[df3.RestingBP<=(df3.RestingBP.mean()+3*df3.RestingBP.std())]
df4.shape
```

```python
le= LabelEncoder()
df4.Sex= le.fit_transform(df4.Sex)
df4.RestingECG= le.fit_transform(df4.RestingECG)
df4.ExerciseAngina= le.fit_transform(df4.ExerciseAngina)
df4.HeartDisease= le.fit_transform(df4.HeartDisease)
df4.ChestPainType= le.fit_transform(df4.ChestPainType)
df4.ST_Slope= le.fit_transform(df4.ST_Slope)
df4.head()
```

```python
X= df4.drop("HeartDisease", axis=1)
Y= df4["HeartDisease"]
```

```python
hscaler= StandardScaler()
X_scaled= hscaler.fit_transform(X)
X_scaled
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.2, stratify=Y, random_state=10)
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
hsvcscores= cross_val_score(SVC(), X_train, Y_train, cv=5)
hsvcscores.mean()
```

```python
svc_bag= BaggingClassifier(estimator=SVC(), n_estimators=100, oob_score=True, max_samples=0.8, random_state=0)
svc_bag.fit(X_train, Y_train)
svc_bag.oob_score_
```

```python
hdstScores= cross_val_score(DecisionTreeClassifier(), X_train, Y_train, cv=5)
hdstScores.mean()
```

```python
dst_bag= BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=100, oob_score=True, max_samples=0.8, random_state=0)
dst_bag.fit(X_train, Y_train)
dst_bag.oob_score_
```
