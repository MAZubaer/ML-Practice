```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn import tree
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("salaries.csv")
df.head()
```

```python
input = df.drop("salary_more_then_100k", axis="columns")
input.head()
```

```python
target = df["salary_more_then_100k"]
target.head()
```

```python
le_comapny = LabelEncoder()
le_job = LabelEncoder()
le_degree = LabelEncoder()
```

```python
input["company"] = le_comapny.fit_transform(input["company"])
input["job"] = le_job.fit_transform(input["job"])
input["degree"] = le_degree.fit_transform(input["degree"])
input.head()
```

```python
tree_model = tree.DecisionTreeClassifier()
x_train,x_test,y_train,y_test = train_test_split(input, target, test_size=0.2, random_state=42)
x_train,x_test,y_train,y_test
```

```python
len(x_train),len(x_test),len(y_train),len(y_test)
```

```python
tree_model = tree.DecisionTreeClassifier()
tree_model.fit(x_train, y_train)
```

```python
tree_model.predict(x_test)
```

```python
tree_model.score(x_test, y_test)
```

```python
df1= pd.read_csv("titanic.csv")
df1.head()
```

```python
df1.drop(["PassengerId", "Name", "SibSp", "Parch","Ticket", "Cabin", "Embarked"], axis="columns", inplace=True)
df1.head()
```

```python
le_sex = LabelEncoder()
df1["Sex"] = le_sex.fit_transform(df1["Sex"])
df1.head()
```

```python
X= df1.drop("Survived", axis="columns")
X.head()
```

```python
Y= df1["Survived"]
Y.head()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
titanic_tree_model = tree.DecisionTreeClassifier()
titanic_tree_model.fit(X_train, Y_train)
```

```python
titanic_tree_model.predict(X_test)
```

```python
titanic_tree_model.score(X_test, Y_test)
```
